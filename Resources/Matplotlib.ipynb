```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Types of plots

### Line plots

```python
x = [1,2,3,4,5]
y= [10,20,25,30,40]
```

```python
plt.plot(x,y)
plt.title('Simple Line Plot')
plt.xlabel('X-axis label')
plt.ylabel('Y-axis label')
plt.grid(True)
```

### Bar chart

```python
months = ['Jan', 'Feb', 'Mar', 'Apr']
revenue = [100, 200, 300, 500]
```

```python
plt.figure(figsize=(10, 6))
plt.bar(months, revenue)
plt.title('Monthly Revenue Bar Chart')
plt.xlabel('Months')
plt.ylabel('Revenue');
```

### Histogram

```python
marks50 = np.random.randint(0, 100, 50)
```

```python
marks50
```

```python
plt.hist(marks50);
plt.xlabel('Marks');
plt.ylabel('No. of Students')
```

```python
plt.hist(marks50, bins = 2)
```

#### Scatter Plot

```python
heights = [65, 70, 68, 72, 66]
weights = [150, 160, 155, 180, 145]
```

```python
plt.scatter(heights, weights)
plt.title('Heights vs. Weights Scatter Plot')
plt.xlabel('Height (inches)')
plt.ylabel('Weight (pounds)')
```

### Subplots and Legends

```python
days = [1,2,3,4,5]
original_price = [10, 20, 25, 30, 40]
predicted_price = [8, 21, 25, 28, 38]
```

```python
plt.plot(days, original_price)
plt.plot(days, predicted_price)
plt.title('Plot with legends')
plt.xlabel('Weekdays')
plt.ylabel('Price')
```
